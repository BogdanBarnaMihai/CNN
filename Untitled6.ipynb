```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader,TensorDataset
from sklearn.model_selection import train_test_split
import torchvision
import copy
import matplotlib.pyplot as plt
from torchsummary import summary

device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
```

```python
def load_mnist(path="/content/sample_data/mnist_train_small.csv"):
  data=np.loadtxt(open(path,'rb'),delimiter=',')
  labels=data[:,0]
  images=data[:,1:]/np.max(data)
  images=images.reshape(-1,1,28,28)
  return images,labels
def make_loaders(images,labels,batchsize=32,test_size=0.1, label_dtype=torch.long):
  dataT=torch.tensor(images).float()
  labelsT=torch.tensor(labels).to(label_dtype)
  train_data,test_data,train_labels,test_labels=train_test_split(dataT,labelsT,test_size=test_size)
  train_loader=DataLoader(TensorDataset(train_data,train_labels),batch_size=batchsize,shuffle=True,drop_last=True)
  test_loader=DataLoader(TensorDataset(test_data,test_labels),batch_size=test_data.shape[0])
  return train_loader,test_loader
```

```python
def generate_gaussians(nPerClass=1000,imgSize=91,widths=(1.8,2.4),noise=0.2):
  x=np.linspace(-4,4,imgSize)
  X,Y=np.meshgrid(x,x)
  images=torch.zeros(2*nPerClass,1,imgSize,imgSize)
  labels=torch.zeros(2*nPerClass)
  for i in range(2*nPerClass):
    ro=2*np.random.randn(2)
    G=np.exp(-((X-ro[0])**2+(Y-ro[1])**2)/(2*widths[i%2]**2))
    G += np.random.randn(imgSize, imgSize) * noise
    images[i, 0] = torch.tensor(G)
    labels[i] = i % 2
  return images, labels[:, None]
```

```python
def generate_gaussian_regression(nGauss=1000, imgSize=91, noise=0.1):
    x = np.linspace(-4, 4, imgSize)
    X, Y = np.meshgrid(x, x)
    images = torch.zeros(nGauss, 1, imgSize, imgSize)
    labels = torch.zeros(nGauss, 3)
    for i in range(nGauss):
        loc = np.max(x) / 2 * np.random.randn(2)
        wid = np.random.rand() * 10 + 5
        G = np.exp(-((X - loc[0])**2 + (Y - loc[1])**2) / wid)
        G += np.random.randn(imgSize, imgSize) * noise
        images[i, 0] = torch.tensor(G)
        labels[i] = torch.tensor([loc[0], loc[1], wid])
    return images, labels
```

```python
def generate_occluded_gaussians(nGauss=1000, imgSize=91):
    x = np.linspace(-4, 4, imgSize)
    X, Y = np.meshgrid(x, x)
    widths = np.linspace(2, 20, nGauss)
    imagesOcc = torch.zeros(nGauss, 1, imgSize, imgSize)
    imagesNoOcc = torch.zeros(nGauss, 1, imgSize, imgSize)
    for i in range(nGauss):
        ro = 1.5 * np.random.randn(2)
        G = np.exp(-((X - ro[0])**2 + (Y - ro[1])**2) / widths[i])
        G += np.random.randn(imgSize, imgSize) / 5
        imagesNoOcc[i, 0] = torch.tensor(G)
        i1 = np.random.choice(np.arange(10, imgSize - 10))
        i2 = np.random.choice(np.arange(2, 6))
        if np.random.randn() > 0:
            G[i1:i1 + i2, :] = 1
        else:
            G[:, i1:i1 + i2] = 1
        imagesOcc[i, 0] = torch.tensor(G)
    return imagesOcc, imagesNoOcc
```

```python
def load_emnist(root='emnist'):
    cdata = torchvision.datasets.EMNIST(root=root, split='letters', download=True)
    letterCategories = cdata.classes[1:]
    images = cdata.data.view([-1, 1, 28, 28]).float()
    images /= torch.max(images)
    labels = copy.deepcopy(cdata.targets) - 1
    return images, labels, letterCategories
```

```python
class MNISTNet(nn.Module):
  def __init__(self,printtoggle=False):
    super().__init__()
    self.print=printtoggle
    self.conv1=nn.Conv2d(1,10,kernel_size=5,stride=1,padding=2)
    self.conv2=nn.Conv2d(10,20,kernel_size=5,stride=1,padding=2)
    self.fc1=nn.Linear(20*7*7,50)
    self.out=nn.Linear(50,10)
  def forward(self,x):
    if self.print: print(f"Input:{x.shape}")
    x=F.relu(F.max_pool2d(self.conv1(x),2))
    if self.print: print(f"conv1:{x.shape}")
    x=F.relu(F.max_pool2d(self.conv2(x),2))
    if self.print: print(f"conv2:{x.shape}")
    x=x.view(x.shape[0],-1)
    x=F.relu(self.fc1(x))
    return self.out(x)
```

```python
class MLPBaseline(nn.Module):
  def __init__(self):
    super().__init__()
    self.net=nn.Sequential(
        nn.Linear(784,256),nn.ReLU(),
        nn.Linear(256,128),nn.ReLU(),
        nn.Linear(128,10)
    )
  def forward(self,x):
    x = x.view(x.size(0), -1)
    return self.net(x)
```

```python
class FlexibleCNN(nn.Module):
  def __init__(self,in_ch=1,out_dim=2,chans=(6,4),kernels=(3,3),fc_units=(50,),img_size=91):
    super().__init__()
    layers,prev_ch,size=[],in_ch,img_size
    for c, k in zip(chans, kernels):
      layers+=[nn.Conv2d(prev_ch,c,k,padding=k//2),nn.ReLU(),nn.AvgPool2d(2,2)]
      size//=2
      prev_ch=c
    self.features=nn.Sequential(*layers)
    flat=size*size*prev_ch
    fc=[]
    for u in fc_units:
      fc+=[nn.Linear(flat,u),nn.ReLU()]
      flat=u
    fc.append(nn.Linear(flat,out_dim))
    self.classifier=nn.Sequential(*fc)
  def forward(self, x):
   return self.classifier(self.features(x).flatten(1))
```

```python
class FeatureMapCNN(nn.Module):
  def __init__(self):
    super().__init__()
    self.conv1=nn.Conv2d(1,6,3,padding=1)
    self.conv2=nn.Conv2d(6,4,3,padding=1)
    self.fc1=nn.Linear(22*22*4,50)
    self.fc2=nn.Linear(50,1)
  def forward(self,x):
    conv1act=F.relu(self.conv1(x))
    x=F.avg_pool2d(conv1act,2)
    conv2act=F.relu(self.conv2(x))
    x=F.avg_pool2d(conv2act,2)
    x=x.view(x.shape[0],-1)
    x=F.relu(self.fc1(x))
    out=self.fc2(x)
    return out
```

```python
class GaussianAutoencoder(nn.Module):
  def __init__(self):
    super().__init__()
    self.env=nn.Sequential(
        nn.Conv2d(1,6,3,padding=1),nn.ReLU(), nn.ReLU(),nn.MaxPool2d(2,2),
        nn.Conv2d(6,4,3,padding=1),nn.ReLU(),nn.ReLU(),nn.MaxPool2d(2,2))
    self.dec=nn.Sequential(
        nn.ConvTranspose2d(4,6,3,stride=2),nn.ReLU(),
        nn.ConvTranspose2d(6,1,3,stride=2))
  def forward(self,x):
    return self.dec(self.env(x))
```

```python
class EMNISTNet(nn.Module):
  def __init__(self):
    super().__init__()
    self.conv1,self.bnorm1=nn.Conv2d(1,64,3,padding=1),nn.BatchNorm2d(64)
    self.conv2,self.bnorm2=nn.Conv2d(64,128,3),nn.BatchNorm2d(128)
    self.conv3,self.bnorm3=nn.Conv2d(128,256,3),nn.BatchNorm2d(256)
    self.fc1=nn.Linear(2*2*256,256)
    self.fc2=nn.Linear(256,64)
    self.fc3=nn.Linear(64,26)
  def forward(self,x):
    x=F.dropout(F.leaky_relu(self.bnorm1(F.max_pool2d(self.conv1(x),2))),0.25,self.training)
    x = F.dropout(F.leaky_relu(self.bnorm2(F.max_pool2d(self.conv2(x), 2))),
                      0.25, self.training)
    x = F.dropout(F.leaky_relu(self.bnorm3(F.max_pool2d(self.conv3(x), 2))),
                  0.25, self.training)
    x = x.view(x.shape[0], -1)
    x = F.dropout(F.leaky_relu(self.fc1(x)), 0.5, self.training)
    x = F.dropout(F.leaky_relu(self.fc2(x)), 0.5, self.training)
    return self.fc3(x)
```

```python
class L1Loss(nn.Module):
    def forward(self, yHat, y):
        return torch.mean(torch.abs(yHat - y))
```

```python
class L2PlusAverageLoss(nn.Module):
    def __init__(self, alpha=1.0):
        super().__init__()
        self.alpha = alpha
    def forward(self, yHat, y):
        return torch.mean((yHat - y)**2) + self.alpha * torch.abs(torch.mean(yHat))
```

```python
class CorrelationLoss(nn.Module):
    def forward(self, yHat, y):
        mx, my = torch.mean(yHat), torch.mean(y)
        num = torch.sum((yHat - mx) * (y - my))
        den = (torch.numel(y) - 1) * torch.std(yHat) * torch.std(y)
        return -num / den
```

```python
def make_optimizer(net,lr=1e-3):
  return torch.optim.Adam(net.parameters(),lr=lr)
```

```python
from torch.optim import optimizer
def train_classifier(net,train_loader,test_loader,numepochs=10,lr=1e-3,verbose=True):
  net=net.to(device)
  lossfun=nn.CrossEntropyLoss()
  optimizer=make_optimizer(net,lr)
  trainloss=torch.zeros(numepochs)
  testLoss  = torch.zeros(numepochs)
  trainAcc  = torch.zeros(numepochs)
  testAcc   = torch.zeros(numepochs)
  for epochi in range(numepochs):
    net.train()
    batchLoss,batchAcc=[],[]
    for x, y in train_loader:
      x,y=x.to(device),y.to(device)
      yhat=net(x)
      loss=lossfun(yhat,y)
      optimizer.zero_grad()
      loss.backward()
      optimizer.step()
      batchLoss.append(loss.item())
      batchAcc.append(100 * torch.mean(
          (torch.argmax(yhat, axis=1) == y).float()).item())
    trainloss[epochi] = np.mean(batchLoss)
    trainAcc[epochi]  = np.mean(batchAcc)
    net.eval()
    with torch.no_grad():
      x,y=next(iter(test_loader))
      x,y=x.to(device),y.to(device)
      yhat=net(x)
      loss=lossfun(yhat,y)
      testLoss[epochi] = loss.item()
      testAcc[epochi]=100*torch.mean((torch.argmax(yhat, axis=1)==y).float()).item()
    if verbose:
      print(f'Epoch {epochi+1:2d}/{numepochs} | '
            f'Train loss {trainloss[epochi]:.4f}, '
            f'acc {trainAcc[epochi]:.2f}% | '
            f'Test loss {testLoss[epochi]:.4f}, '
            f'acc {testAcc[epochi]:.2f}%')
  return trainloss, testLoss, trainAcc, testAcc, net
```

```python
def train_regressor(net,train_loader,test_loader,numepochs=30,lr=1e-3,verbose=True):
  net=net.to(device)
  lossfun=nn.MSELoss()
  optimizer=make_optimizer(net,lr)
  trainLoss=torch.zeros(numepochs)
  testLoss=torch.zeros(numepochs)
  for epochi in range(numepochs):
    net.train()
    batchLoss=[]
    for x, y in train_loader:
      x,y=x.to(device),y.to(device)
      yhat=net(x)
      loss=lossfun(yhat,y)
      optimizer.zero_grad()
      loss.backward()
      optimizer.step()
      batchLoss.append(loss.item())
    trainLoss[epochi]=np.mean(batchLoss)
    net.eval()
    with torch.no_grad():
      x,y=next(iter(test_loader))
      x,y=x.to(device),y.to(device)
      testLoss[epochi]=lossfun(net(x),y).item()
    if verbose and(epochi%5==0 or epochi==numepochs-1):
      print(f'Epoch {epochi+1:3d}/{numepochs} | 'f'Train {trainLoss[epochi]:.4f} | Test {testLoss[epochi]:.4f}')
  return trainLoss, testLoss, net
```

```python
def train_autoencoder(net,images,numepochs=500,lr=1e-3,batch_size=32,target=None,lossfn=None,verbose=True):
  net=net.to(device)
  if lossfn is None:
    lossfn=nn.MSELoss()
  optimizer=make_optimizer(net,lr)
  losses=torch.zeros(numepochs)
  n=images.shape[0]
  if target is None:
    target=images
  for epochi in range(numepochs):
    idx=np.random.choice(n,size=batch_size,replace=False)
    x=images[idx].to(device)
    y=target[idx].to(device)
    yhat=net(x)
    loss=lossfn(yhat,y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    losses[epochi]=loss.item()
    if verbose and (epochi%100==0 or epochi== numepochs-1):
      print(f'Epoch {epochi:4d}/{numepochs} | Loss {losses[epochi]:.5f}')
  return losses,net
```

```python
def plot_training_curves(trainLoss,testLoss,trainAcc,testAcc,title_suffix='',figsize=(16,5)):
  fig,ax=plt.subplots(1,2,figsize=figsize)
  ax[0].plot(trainLoss,label='Train')
  ax[0].plot(testLoss,label='Test')
  ax[0].set_xlabel('Epochs'); ax[0].set_ylabel('Loss')
  ax[0].set_title('Model loss'); ax[0].legend()
  ax[1].plot(trainAcc, 's-', label='Train')
  ax[1].plot(testAcc,  'o-', label='Test')
  ax[1].set_xlabel('Epochs'); ax[1].set_ylabel('Accuracy (%)')
  ax[1].set_title(f'Final test acc: {testAcc[-1]:.2f}% {title_suffix}')
  ax[1].legend()
  plt.tight_layout(); plt.show()
```

```python
def plot_gaussians(images, labels=None, n=21, cmap='jet'):
  rows=int(np.ceil(n/7))
  fig,axes=plt.subplots(rows,7,figsize=(13,2*rows))
  for i,ax in enumerate(np.array(axes).flatten()):
    if i>=n: axes.axis('off'); continue
    G=torch.squeeeze(images[i,0]).detach().cpu()
    ax.imshow(G,vmin=-1,vmax=1,cmap=cmap)
    if labels is not None:
      ax.set_title(f"Class {int(labels[i])}")
    ax.axis('off')
  plt.tight_layout(); plt.show()
```

```python
def plot_feature_maps(featmap,x,n_images=10,title='feature map activations'):
  n_maps=featmap.shape[1]
  fig,axs=plt.subplots(n_maps + 1, n_images, figsize=(13, 2 * (n_maps + 1)))
  for pici in range(n_images):
    img=x[pici,0].detach().cpu()
    axs[0, pici].imshow(img, cmap='jet', vmin=0, vmax=1)
    axs[0, pici].axis('off')
    if pici == 0:
        axs[0, pici].set_ylabel('Original', rotation=0, ha='right', va='center')
    for feati in range(n_maps):
      fm=featmap[pici,feati].detach().cpu()
      axs[feati + 1, pici].imshow(fm, cmap='inferno', vmin=0, vmax=torch.max(fm) * 0.9)
      axs[feati + 1, pici].axis('off')
      if pici == 0:
        axs[feati + 1, pici].set_ylabel(f'FM {feati}', rotation=0, ha='right', va='center')
  plt.tight_layout()
  plt.suptitle(title, x=0.5, y=1.02)
  plt.show()
```

```python
def plot_filter_correlations(featmap):
  nStim,nMaps=featmap.shape[0],featmap.shape[1]
  nCors=(nMaps*(nMaps-1))//2
  allrs=np.zeros((nStim,nCors))
  Call=np.zeros((nMaps,nMaps))
  for i in range(nStim):
    fm=featmap[i].view(nMaps,-1).detach().cpu()
    C=np.corrcoef(fm)
    Call+=C
    idx=np.nonzero(np.triu(C,1))
    allrs[i,:]=C[idx]
  fig = plt.figure(figsize=(14, 5))
  ax0 = fig.add_axes([0.10, 0.10, 0.55, 0.85])
  ax1 = fig.add_axes([0.68, 0.10, 0.30, 0.85])
  cax = fig.add_axes([0.99, 0.10, 0.01, 0.85])
  for i in range(nCors):
    ax0.plot(i + np.random.randn(nStim) / 30, allrs[:, i],'o', markerfacecolor='w', markersize=8)
  ax0.set_xlim([-0.5, nCors - 0.5]); ax0.set_ylim([-1.05, 1.05])
  ax0.set_xticks(range(nCors))
  ax0.set_xticklabels([f'{idx[0][i]}-{idx[1][i]}' for i in range(nCors)])
  ax0.set_xlabel('Feature map pair'); ax0.set_ylabel('Correlation')
  ax0.set_title('Correlations per image')
  h = ax1.imshow(Call / nStim, vmin=-1, vmax=1)
  ax1.set_title('Mean correlation matrix')
  ax1.set_xlabel('Feature map'); ax1.set_ylabel('Feature map')
  fig.colorbar(h, cax=cax); plt.show()
```

```python
import sklearn.metrics as skm

def plot_confusion_matrix(y_true, y_pred, categories):
    C = skm.confusion_matrix(y_true, y_pred, normalize='true')
    fig, ax = plt.subplots(figsize=(10, 10))
    ax.imshow(C, 'Blues', vmax=0.05)
    ax.set_xticks(range(len(categories)))
    ax.set_yticks(range(len(categories)))
    ax.set_xticklabels(categories); ax.set_yticklabels(categories)
    ax.set_xlabel('Predicted'); ax.set_ylabel('True')
    ax.set_title('Test confusion matrix')
    plt.show()
    return C
```

```python
def exp1_mnist_baseline():
  print(f"Experiment 1: MNIST baseline")
  images,labels=load_mnist()
  train_loader,test_loader=make_loaders(images,labels)
  net=MNISTNet()
  net = net.to(device)
  summary(net,(1,28,28))
  trainLoss,testLoss,trainAcc,testAcc,net=train_classifier(net,train_loader,test_loader,numepochs=10)
  plot_training_curves(trainLoss,testLoss,trainAcc,testAcc)
  return net,testAcc[-1].item()
```

```python
def exp2_translation_invariance():
  print(f"Experiment 2: Translation invariance(CNN vs MLP)")
  images,labels=load_mnist()
  train_loader,test_loader=make_loaders(images,labels,batchsize=32)
  cnn=MNISTNet()
  _,_,_,cnnAcc,cnn=train_classifier(cnn,train_loader,test_loader,numepochs=10,verbose=False)
  mlp=MLPBaseline()
  _,_,_,mlpAcc,mlp=train_classifier(mlp,train_loader,test_loader,numepochs=10,verbose=False)
  shifts=[0,2,4,6,8,10]
  cnn_shift,mlp_shift=[],[]
  X_test,y_test=next(iter(test_loader))
  for s in shifts:
    Xs=torch.roll(X_test,shifts=s,dims=3)
    with torch.no_grad():
      cnn_pred=cnn(Xs.to(device)).argmax(1).cpu()
      mlp_pred=mlp(Xs.to(device)).argmax(1).cpu()
      cnn_shift.append(100*torch.mean((cnn_pred==y_test).float()).item())
      mlp_shift.append(100*torch.mean((mlp_pred==y_test).float()).item())
  fig, ax = plt.subplots(figsize=(8, 5))
  ax.plot(shifts, cnn_shift, 's-', label='CNN', linewidth=2)
  ax.plot(shifts, mlp_shift, 'o-', label='MLP', linewidth=2)
  ax.set_xlabel('Shift (pixels)')
  ax.set_ylabel('Test accuracy (%)')
  ax.set_title('Translation invariance: CNN vs MLP')
  ax.legend(); ax.grid(True)
  plt.show()
  return cnn_shift, mlp_shift
```

```python
def exp3_gaussian_regression():
  print(f"Experiment 3: Gaussian regression")
  images,labels=generate_gaussian_regression(nGauss=1000,imgSize=91)
  train_loader,test_loader=make_loaders(images,labels,batchsize=16, label_dtype=torch.float)
  net=FlexibleCNN(in_ch=1,out_dim=3,chans=(6,4),kernels=(3,3),fc_units=(50,),img_size=91)
  trainLoss,testLoss,net=train_regressor(net,train_loader,test_loader,numepochs=30)
  net.eval()
  with torch.no_grad():
    X,y=next(iter(test_loader))
    X=X.to(device)
    yHat=net(X).cpu()
  fig, ax = plt.subplots(figsize=(6, 6))
  for i, name in enumerate(['Cx', 'Cy', 'width']):
      cr = np.corrcoef(y[:, i], yHat[:, i])[0, 1]
      ax.plot(y[:, i], yHat[:, i], 'o',
              label=f'{name}, r={cr:.3f}', alpha=0.5)
  ax.legend(); ax.set_xlabel('True'); ax.set_ylabel('Predicted')
  ax.set_title('Gaussian regression correlations')
  ax.grid(True); plt.show()
  return net
```

```python
def exp4_hyperparamter_sweep():
  print(f"Experiment 4: Hyperparameter sweep")
  images,labels=generate_gaussian_regression(nGauss=600,imgSize=91)
  train_loader,test_loader=make_loaders(images,labels,batchsize=16, label_dtype=torch.float)
  convChans=[2,5,8]
  results=np.zeros((len(convChans),len(convChans),2))
  for i,Ni in enumerate(convChans):
    for j,Nj in enumerate(convChans):
      net=FlexibleCNN(in_ch=1,out_dim=3,chans=(Ni,Nj),kernels=(3,3),fc_units=(50,),img_size=91)
      _,_,net=train_regressor(net,train_loader,test_loader,numepochs=30,verbose=False)
      results[i,j,0]=Ni+Nj
      with torch.no_grad():
          x,y=next(iter(test_loader))
          yhat=net(x.to(device)).cpu()
          correlation = np.corrcoef(y[:, 0].numpy(), yhat[:, 0].numpy())[0, 1]
      results[i, j, 1] = correlation
      print(f'Chans=({Ni},{Nj}) -> test correlation {correlation:.2f}')
  fig, ax = plt.subplots(1, 2, figsize=(12, 4))
  im = ax[0].imshow(results[:, :, 1], vmin=results[:, :, 1].min(),vmax=results[:, :, 1].max())
  ax[0].set_xticks(range(len(convChans))); ax[0].set_yticks(range(len(convChans)))
  ax[0].set_xticklabels(convChans); ax[0].set_yticklabels(convChans)
  ax[0].set_xlabel('Channels in conv1'); ax[0].set_ylabel('Channels in conv2')
  ax[0].set_title('Test Correlation')
  plt.colorbar(im, ax=ax[0])
  ax[1].scatter(results[:, :, 0].flatten(), results[:, :, 1].flatten())
  ax[1].set_xlabel('Total conv channels')
  ax[1].set_ylabel('Test Correlation')
  ax[1].set_title('Diminishing returns')
  ax[1].grid(True)
  plt.tight_layout(); plt.show()
  fcUnits=np.round(np.linspace(5,500,10)).astype(int)
  fcAcc=[]
  for u in fcUnits:
    net=FlexibleCNN(in_ch=1,out_dim=3,chans=(6,4),kernels=(3,3),fc_units=(u,),img_size=91)
    _,_,net=train_regressor(net,train_loader,test_loader,numepochs=30,verbose=False)
    with torch.no_grad():
        x,y=next(iter(test_loader))
        yhat=net(x.to(device)).cpu()
        correlation = np.corrcoef(y[:, 0].numpy(), yhat[:, 0].numpy())[0, 1]
    fcAcc.append(correlation)
    print(f'FC units={u} -> test correlation {correlation:.2f}')
  plt.figure(figsize=(8, 5))
  plt.plot(fcUnits, fcAcc, 's-')
  plt.xlabel('Number of units in final FC layer')
  plt.ylabel('Test Correlation')
  plt.title('FC width vs. correlation')
  plt.grid(True); plt.show()
```

```python
def exp8_emnist_letters():
  print(f"Experiment 8: EMNIST letters")
  images,labels,letterCategories=load_emnist()
  train_loader,test_loader=make_loaders(images,labels,batchsize=32,test_size=0.05)
  net=EMNISTNet()
  trainLoss,testLoss,trainErr,testErr=train_with_error(net,train_loader,test_loader,numepochs=10)
  fig,ax=plt.subplots(1,2,figsize=(16,5))
  ax[0].plot(trainLoss,'s-',label='Train')
  ax[0].plot(testLoss,'o-',label='Test')
  ax[0].set_xlabel('Epochs'); ax[0].set_ylabel('Loss')
  ax[0].set_title('Model loss');ax[0].legend()
  ax[1].plot(trainErr,'s-',label='Train')
  ax[1].plot(testErr,'o-',label='Test')
  ax[1].set_xlabel('Epochs'); ax[1].set_ylabel('Error (%)')
  ax[1].set_title(f'Final test error {testErr[-1]:.2f}%')
  ax[1].legend()
  plt.tight_layout(); plt.show()
  x,y=next(iter(test_loader))
  with torch.no_grad():
    yhat=net(x.to(device)).argmax(1).cpu()
  plot_confusion_matrix(y.numpy(),yhat.numpy(),letterCategories)
  return net
```

```python
def train_with_error(net,train_loader,test_loader,numepochs=10,lr=1e-3):
  net=net.to(device)
  lossfun=nn.CrossEntropyLoss()
  optimizer=make_optimizer(net,lr)
  trainLoss,testLoss=torch.zeros(numepochs),torch.zeros(numepochs)
  trainErr,testErr=torch.zeros(numepochs),torch.zeros(numepochs)
  for epochi in range(numepochs):
    net.train()
    bL,bE=[],[]
    for x,y in train_loader:
      x,y=x.to(device),y.to(device)
      yhat=net(x)
      loss=lossfun(yhat,y)
      optimizer.zero_grad(); loss.backward(); optimizer.step()
      bL.append(loss.item())
      bE.append(100*(yhat.argmax(1)!=y).float().mean().item())
    trainLoss[epochi],trainErr[epochi]=np.mean(bL),np.mean(bE)
    net.eval()
    with torch.no_grad():
      x,y=next(iter(test_loader))
      x,y=x.to(device),y.to(device)
      yhat=net(x)
      testLoss[epochi]=lossfun(yhat,y).item()
      testErr[epochi]=100*(yhat.argmax(1)!=y).float().mean().item()
    print(f'Epoch {epochi+1:2d}/{numepochs} | '
              f'Train err {trainErr[epochi]:.2f}% | '
              f'Test err {testErr[epochi]:.2f}%')
  return trainLoss, testLoss, trainErr, testErr
```
